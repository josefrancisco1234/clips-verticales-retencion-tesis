# Resultados del selector de clips: matriz de confusión, ROC y precisión-exhaustividad
Tesis de J. F. Núñez del Prado Vega (U. de Lima). Ventanas de 640 videos no usados para fijar la fórmula; clase *caliente* = ventana que toca un pico de Most Replayed (solo califica, nunca es entrada).

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import confusion_matrix, classification_report, roc_curve, auc, precision_recall_curve, average_precision_score
d = np.load('reports/ventanas_eval_D099.npz')
y, s, p = d['y_true'], d['score'], d['elegida']
print(len(y), 'ventanas;', y.sum(), 'calientes')
```

## Reporte de clasificación

```python
print(classification_report(y, p, target_names=['No caliente', 'Caliente'], digits=3))
```

## Matriz de confusión

```python
cm = confusion_matrix(y, p, labels=[True, False])
fig, ax = plt.subplots(figsize=(5.2, 4.2))
im = ax.imshow(cm, cmap='Blues')
for i in range(2):
    for j in range(2):
        ax.text(j, i, f'{cm[i, j]:,}', ha='center', va='center', fontweight='bold', fontsize=13, color='white' if cm[i, j] > cm.max()/2 else 'black')
ax.set_xticks([0, 1]); ax.set_xticklabels(['Elegida', 'No elegida'])
ax.set_yticks([0, 1]); ax.set_yticklabels(['Caliente', 'No caliente'], rotation=90, va='center')
ax.set_xlabel('Clase predicha'); ax.set_ylabel('Clase real'); ax.set_title('Matriz de confusión')
fig.colorbar(im); plt.tight_layout(); plt.show()
```

## Curvas ROC y precisión-exhaustividad

```python
fpr, tpr, _ = roc_curve(y, s); pr, rc, _ = precision_recall_curve(y, s)
fig, ax = plt.subplots(1, 2, figsize=(9, 3.6))
ax[0].plot(fpr, tpr, label=f'Fórmula V1 (AUC = {auc(fpr, tpr):.3f})'); ax[0].plot([0, 1], [0, 1], '--', c='gray', label='Azar')
ax[0].set_xlabel('Tasa de falsos positivos'); ax[0].set_ylabel('Tasa de verdaderos positivos'); ax[0].set_title('Curva ROC'); ax[0].legend()
ax[1].plot(rc, pr, label=f'Fórmula V1 (AP = {average_precision_score(y, s):.3f})'); ax[1].axhline(y.mean(), ls='--', c='gray', label='Azar')
ax[1].set_xlabel('Exhaustividad'); ax[1].set_ylabel('Precisión'); ax[1].set_title('Precisión-exhaustividad'); ax[1].legend()
plt.tight_layout(); plt.show()
```
